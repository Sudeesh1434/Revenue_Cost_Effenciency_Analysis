```python
# Created column by Revenue & Total cost
df['Calculated_Profit']=df['Revenue']-df['Total_Cost']
df
```

```python
# checked if Profit is matched with created column
(df['Profit']==df['Calculated_Profit']).value_counts()
```

```python
# Created column by Profit & Revenue
df['Calculated_Profit_Margin_%']=(df['Profit']/df['Revenue'])*100
df
```

```python
# Aggregate Revenue & Profit by Product
product_summary=df.groupby('Product')[['Revenue','Profit']].sum()
product_summary
```

```python
# Add Profit Margin at Product Level
product_summary['Profit_Margin_%']=(product_summary['Profit']/product_summary['Revenue'])*100
product_summary
```

```python
# Identify High-Revenue, Low-Profit Products
product_summary.sort_values(['Revenue','Profit'],ascending=[False,True])
```

```python
# Aggregate costs by product
cost_summary=df.groupby('Product')[['COGS','Operating_Cost','Marketing_Cost','Total_Cost','Revenue']].sum()
cost_summary
```

```python
# Convert costs into percentages
cost_summary['COGS%']=(cost_summary['COGS']/cost_summary['Revenue'])*100
cost_summary['Operating_Cost%']=(cost_summary['Operating_Cost']/cost_summary['Revenue'])*100
cost_summary['Marketing_Cost_%']=(cost_summary['Marketing_Cost']/cost_summary['Revenue'])*100
cost_summary
```

```python
# Aggregate revenue & profit by region
region_summary=df.groupby('Region')[['Revenue','Profit']].sum()
region_summary
```

```python
# Calculate profit margin by region
region_summary['Profit_Margin_%']=(region_summary['Profit']/region_summary['Revenue'])
region_summary
```
